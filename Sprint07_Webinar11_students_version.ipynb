## Ejercicio 8 · Outliers con IQR \(8 min\)

La regla IQR permite detectar valores inusualmente altos o bajos.

No significa que esos valores estén mal. Significa que deben revisarse antes de sacar conclusiones.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# TODO estudiante: escribe aquí tu código.
# Función para calcular límites IQR
def calcular_IQR(tabla, columna):
    q1 = tabla[columna].quantile(0.25)
    q3 = tabla[columna].quantile(0.75)
    IQR = q3 - q1
    limite_inferior = q1 - 1.5 * IQR
    limite_superior = q3 + 1.5 * IQR
    return limite_inferior, limite_superior
````

````python
inferior, superior = calcular_IQR(df_limpio, 'ingreso')
````

````python
# Indentificar Ventas Atípicas por Ingreso
print('Limite Inferior', inferior)
print('Limite Superior', superior)
````

````python
outliers = df_limpio[
    (df_limpio['ingreso']<inferior) |
    (df_limpio['ingreso']>superior)
]

print('Cantidad de posibles atipicos:', len(outliers))
outliers[['venta_id', 'pais','canal', 'categoria', 'ingreso', 'margen']]
````

````python
# Función para segmentación
def clasificador(ingreso):
    if ingreso >= 4_000_000:
        return 'Alta'
    elif ingreso >= 1_500_000:
        return 'Media'
    else:
        return 'Baja'
````

````python
# Conteo por Segmento
df_limpio['segmento_venta'] = df_limpio['ingreso'].apply(clasificador)
df_limpio[['ingreso','segmento_venta']]
````

````python
# Tabla final para recomendación
ranking = (
    df_limpio
    .groupby(['pais','canal','categoria'], as_index=False)
    .agg(
        ingreso_total = ('ingreso','sum'),
        margen_total = ('margen','sum'),
        ventas = ('venta_id','count')
    )
)

ranking['margen_pct'] = ranking['margen_total'] / ranking['ingreso_total']
rankin = ranking.sort_values('ingreso_total', ascending=False)

ranking.head(10)
````
